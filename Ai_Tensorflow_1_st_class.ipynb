```python
# Topic 1: What is Gradient?

#

# Function: y = x²

# Derivative: dy/dx = 2x

#

# At x = 3:

# dy/dx = 2 × 3 = 6



import tensorflow as tf

x = tf.Variable(3.0)

with tf.GradientTape() as tape:

    y = x **3 + x**2

gradient = tape.gradient(y, x)

print("x =", x.numpy())

print("y = x² =", y.numpy())

print("dy/dx =", gradient.numpy())
```

```python
# Topic: Weight Update Using Gradient Descent

#

# Formula:

# w_new = w_old - learning_rate × dL/dw

#

# Loss function:

# L = (w - target)²



import tensorflow as tf

# Step 1: Define the initial weight

w = tf.Variable(5.0)

# Step 2: Define the target value

target = tf.constant(3.0)

# Step 3: Define the learning rate (η)

learning_rate = 0.1

# Step 4: Record the loss calculation

with tf.GradientTape() as tape:

    # Loss function: L = (w - target)²

    loss = (w - target) ** 2

# Step 5: Calculate dL/dw

gradient = tape.gradient(loss, w)

# Save the old weight only for displaying it

old_weight = w.numpy()

# Step 6: Apply the gradient descent formula

# w_new = w_old - learning_rate × gradient

w.assign_sub(learning_rate * gradient)

print("Old weight:", old_weight)

print("Target:", target.numpy())

print("Loss:", loss.numpy())

print("Gradient dL/dw:", gradient.numpy())

print("New weight:", w.numpy())
```
